```python
import pandas as pd

df = pd.read_csv("lamborghini_sales_2020_2025.csv")

df.head()
```

```python
df_original = df.copy()

df_original.shape
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

df.shape
```

```python
df.isnull().sum()
```

```python
df = df.dropna()

df.shape
```

```python
df.dtypes
```

```python
text_columns = ["Model", "Region", "Color", "Fuel Type", "Turbo (Yes/No)"]

for col in text_columns:
    df[col] = df[col].str.strip().str.lower()

df.head()
```

```python
df["Turbo (Yes/No)"] = df["Turbo (Yes/No)"].map({"yes": 1, "no": 0})

df["Turbo (Yes/No)"].head()
```

```python
df.shape

df.head()
```

```python
df.to_csv("lamborghini_sales_2020_2025_limpio.csv", index=False)
```

```python
pivot = pd.pivot_table(df, values="Sales Volume", index="Model", aggfunc="mean")

pivot
```
